```python
import pandas as pd
import re

# =========================================
# LOAD DATASET
# =========================================

df = pd.read_csv("csv/dataset_pretext.csv")

# =========================================
# CLEAN TEXT
# =========================================

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_text"] = df["pretext"].apply(clean_text)

# =========================================
# PERSONA RULES
# =========================================

PERSONA_RULES = {

    "beginner_explorer": [
        "baru mulai",
        "pemula",
        "beginner",
        "baru belajar",
        "dasar",
    ],

    "path_confused_learner": [
        "bingung",
        "mulai dari mana",
        "roadmap",
        "arah karir",
        "gak tau",
    ],

    "project_seeker": [
        "project",
        "portfolio",
        "portofolio",
        "projek",
    ],

    "skill_gap_aware": [
        "skill",
        "kurang kemampuan",
        "belum bisa",
        "gap",
    ],

    "career_switcher": [
        "switch career",
        "pindah karir",
        "transisi karir",
        "beralih",
    ],

    "validation_seeker": [
        "apakah",
        "worth it",
        "terlambat",
        "masih bisa",
        "ragu",
    ],

    "overwhelmed_learner": [
        "kewalahan",
        "overwhelmed",
        "stress",
        "burnout",
    ]
}

# =========================================
# DETECT PERSONA
# =========================================

all_personas = []

for text in df["clean_text"]:

    for persona, keywords in PERSONA_RULES.items():

        for keyword in keywords:

            if keyword in text:
                all_personas.append(persona)
                break

# =========================================
# AMBIL UNIQUE PERSONA
# =========================================

unique_personas = sorted(list(set(all_personas)))

persona_df = pd.DataFrame({
    "persona_type": unique_personas
})

# =========================================
# SAVE CSV
# =========================================

persona_df.to_csv("persona_taxonomy.csv", index=False)

print(persona_df)
print("\nSaved to persona_taxonomy.csv")
```
