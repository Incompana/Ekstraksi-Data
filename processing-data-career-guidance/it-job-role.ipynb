````python
def normalize_role(title):

    title = str(title).lower()


    # =====================================================
    # DATA
    # =====================================================

    if any(keyword in title for keyword in [

        "data scientist",
        "data science",
        "data analyst",
        "analytics",
        "business intelligence",
        "bi developer",

    ]):
        return "Data Science"


    # =====================================================
    # DATA ENGINEERING
    # =====================================================

    elif any(keyword in title for keyword in [

        "big data",
        "data engineer",
        "etl",
        "hadoop",
        "spark",

    ]):
        return "Data Engineering"


    # =====================================================
    # MACHINE LEARNING / AI
    # =====================================================

    elif any(keyword in title for keyword in [

        "machine learning",
        "artificial intelligence",
        "ai ",
        "ml ",
        "deep learning",

    ]):
        return "AI/ML"


    # =====================================================
    # FRONTEND
    # =====================================================

    elif any(keyword in title for keyword in [

        "frontend",
        "front-end",
        "ui developer",

    ]):
        return "Frontend Development"


    # =====================================================
    # BACKEND
    # =====================================================

    elif any(keyword in title for keyword in [

        "backend",
        "back-end",
        "api developer",

    ]):
        return "Backend Development"


    # =====================================================
    # FULLSTACK
    # =====================================================

    elif any(keyword in title for keyword in [

        "fullstack",
        "full stack",

    ]):
        return "Fullstack Development"


    # =====================================================
    # MOBILE
    # =====================================================

    elif any(keyword in title for keyword in [

        "android",
        "ios",
        "mobile",

    ]):
        return "Mobile Development"


    # =====================================================
    # DEVOPS
    # =====================================================

    elif any(keyword in title for keyword in [

        "devops",
        "ansible",
        "ci/cd",
        "bamboo",

    ]):
        return "DevOps"


    # =====================================================
    # CLOUD
    # =====================================================

    elif any(keyword in title for keyword in [

        "aws",
        "azure",
        "cloud",
        "solutions architect",

    ]):
        return "Cloud Computing"


    # =====================================================
    # CYBERSECURITY
    # =====================================================

    elif any(keyword in title for keyword in [

        "security",
        "cybersecurity",
        "penetration",
        "ethical hacking",

    ]):
        return "Cybersecurity"


    # =====================================================
    # UI UX
    # =====================================================

    elif any(keyword in title for keyword in [

        "ui",
        "ux",
        "accessibility",

    ]):
        return "UI/UX Design"


    # =====================================================
    # BLOCKCHAIN
    # =====================================================

    elif any(keyword in title for keyword in [

        "blockchain",

    ]):
        return "Blockchain"


    # =====================================================
    # NETWORKING
    # =====================================================

    elif any(keyword in title for keyword in [

        "network",

    ]):
        return "Networking"


    # =====================================================
    # PROJECT MANAGEMENT
    # =====================================================

    elif any(keyword in title for keyword in [

        "project manager",
        "agile",

    ]):
        return "Project Management"


    # =====================================================
    # SOFTWARE ENGINEERING
    # =====================================================

    elif any(keyword in title for keyword in [

        "software engineer",
        "software developer",

    ]):
        return "Software Engineering"


    # =====================================================
    # OTHER
    # =====================================================

    return "Other"


# =========================================================
# CREATE NORMALIZED ROLE
# =========================================================

df["normalized_role"] = (
    df["Job Title"]
    .apply(normalize_role)
)

print("\nNormalized roles:\n")

print(
    df["normalized_role"]
    .value_counts()
)


# =========================================================
# CLEAN ROLE DATA
# =========================================================

df["Job Title"] = (
    df["Job Title"]
    .fillna("")
    .astype(str)
    .str.lower()
)

df["Skills"] = (
    df["Skills"]
    .fillna("")
    .astype(str)
    .str.lower()
)


# =========================================================
# ROLE TEXT
# =========================================================

df["role_text"] = (

    df["Job Title"] + " " +
    df["Skills"] + " " +
    df["normalized_role"]

)


# =========================================================
# CLEAN CAREER DATA
# =========================================================

career_df["title"] = (
    career_df["title"]
    .fillna("")
    .astype(str)
    .str.lower()
)

career_df["body"] = (
    career_df["body"]
    .fillna("")
    .astype(str)
    .str.lower()
)


# =========================================================
# CAREER TEXT
# =========================================================

career_df["career_text"] = (

    career_df["title"] + " " +
    career_df["body"]

)


# =========================================================
# LOAD MODEL
# =========================================================

print("\nLoading model...")

model = SentenceTransformer(
    "sentence-transformers/all-mpnet-base-v2"
)


# =========================================================
# ENCODE ROLES
# =========================================================

print("\nEncoding roles...")

role_embeddings = model.encode(

    role_df["role_text"].tolist(),

    batch_size=32,
    show_progress_bar=True,

    normalize_embeddings=True

)


# =========================================================
# ENCODE CAREER POSTS
# =========================================================

print("\nEncoding career posts...")

career_embeddings = model.encode(

    career_df["career_text"].tolist(),

    batch_size=64,
    show_progress_bar=True,

    normalize_embeddings=True

)


# =========================================================
# ROLE MATCHING
# =========================================================

predicted_roles = []
role_scores = []

top3_roles = []
top3_scores = []

print("\nMatching roles...")

for career_emb in tqdm(career_embeddings):

    # =====================================================
    # SIMILARITY
    # =====================================================

    sims = cosine_similarity(
        [career_emb],
        role_embeddings
    )[0]


    # =====================================================
    # TOP ROLE
    # =====================================================

    best_idx = np.argmax(sims)

    best_role = role_df.iloc[best_idx][
        "normalized_role"
    ]

    best_score = sims[best_idx]


    # =====================================================
    # TOP 3
    # =====================================================

    top_indices = np.argsort(sims)[-3:][::-1]

    top_roles = [

        role_df.iloc[i]["normalized_role"]

        for i in top_indices

    ]

    top_scores = [

        round(float(sims[i]), 4)

        for i in top_indices

    ]


    # =====================================================
    # THRESHOLD
    # =====================================================

    if best_score < 0.45:

        best_role = "Unknown"


    # =====================================================
    # SAVE
    # =====================================================

    predicted_roles.append(best_role)

    role_scores.append(
        round(float(best_score), 4)
    )

    top3_roles.append(top_roles)

    top3_scores.append(top_scores)


# =========================================================
# SAVE RESULT
# =========================================================

career_df["predicted_role"] = predicted_roles

career_df["role_score"] = role_scores

career_df["top3_roles"] = top3_roles

career_df["top3_scores"] = top3_scores


# =========================================================
# PREVIEW
# =========================================================

preview_cols = [

    "title",
    "predicted_role",
    "role_score",
    "top3_roles"

]

print("\nTOP RESULTS:\n")

print(
    career_df[preview_cols]
    .head(20)
)


# =========================================================
# SAVE CSV
# =========================================================

output_path = "career_role_prediction.csv"

career_df.to_csv(

    output_path,

    index=False,
    encoding="utf-8-sig"

)

print("\nDONE!")
print("Saved to:", output_path)
```
````
